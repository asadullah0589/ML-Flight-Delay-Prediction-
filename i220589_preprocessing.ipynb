```python
import pandas as pd

def merge_and_show_matches(df_flights, df_weather):
    """
    Merges flight and weather DataFrames and shows matching row numbers
    """
   
    df_flights['Date'] = pd.to_datetime(df_flights['Date'])
    df_weather['TIME'] = pd.to_datetime(df_weather.iloc[:, 0])
    
    
    weather_dates = dict(enumerate(df_weather['TIME']))
    
    
    matches = df_flights['Date'].isin(df_weather['TIME'])
    matching_indices = df_flights[matches].index
    
    
    for flight_idx in matching_indices:
        flight_date = df_flights.loc[flight_idx, 'Date']
        weather_idx = df_weather[df_weather['TIME'] == flight_date].index[0]
        print(f"Flight Row {flight_idx + 1} matches Weather Row {weather_idx + 1}")
    
   
    merged_df = pd.merge(
        df_flights,
        df_weather,
        left_on='Date',
        right_on='TIME',
        how='left'
    )
    
    return merged_df


test_merge = merge_and_show_matches(test_df2, weather_df)
```

```python
import pandas as pd
import os

def process_excel_file(file_path, year):
    # Read excel file
    df = pd.read_excel(file_path, header=None)
    
    #
    month = df.iloc[1, 0].split()[0]  
    days = df.iloc[1, 0].split()[1:]   
    
    
    dates = [f"{year} {month} {day}" for day in days]
    
    
    def process_column(col_data):
        values = col_data.split()[3:] 
        return {
            'Max': values[0::3],
            'Average': values[1::3],
            'Min': values[2::3]
        }
    
    
    measurements = {
        'TIME': dates,
        'Temperature': process_column(df.iloc[1, 1]),
        'Dew_Point': process_column(df.iloc[1, 2]),
        'Humidity': process_column(df.iloc[1, 3]),
        'Wind_Speed': process_column(df.iloc[1, 4]),
        'Pressure': process_column(df.iloc[1, 5])
    }
    
    
    precip_values = df.iloc[1, 6].split()[1:]
    
    
    result_data = {
        'TIME': measurements['TIME'],
        'Max_Temperature': measurements['Temperature']['Max'],
        'Avg_Temperature': measurements['Temperature']['Average'],
        'Min_Temperature': measurements['Temperature']['Min'],
        'Max_Dew_Point': measurements['Dew_Point']['Max'],
        'Avg_Dew_Point': measurements['Dew_Point']['Average'],
        'Min_Dew_Point': measurements['Dew_Point']['Min'],
        'Max_Humidity': measurements['Humidity']['Max'],
        'Avg_Humidity': measurements['Humidity']['Average'],
        'Min_Humidity': measurements['Humidity']['Min'],
        'Max_Wind_Speed': measurements['Wind_Speed']['Max'],
        'Avg_Wind_Speed': measurements['Wind_Speed']['Average'],
        'Min_Wind_Speed': measurements['Wind_Speed']['Min'],
        'Max_Pressure': measurements['Pressure']['Max'],
        'Avg_Pressure': measurements['Pressure']['Average'],
        'Min_Pressure': measurements['Pressure']['Min'],
        'Total_Precipitation': precip_values
    }
    
    
    result_df = pd.DataFrame(result_data)
    
    
    directory = os.path.dirname(file_path)
    file_number = os.path.splitext(os.path.basename(file_path))[0]
    
    
    output_path = os.path.join(directory, f'processed{file_number}.csv')
    result_df.to_csv(output_path, index=False)
    print(f"Saved processed file to: {output_path}")
    
    return result_df

folder_path = "C:\\Users\\swiss computer\\Downloads\\kishti\\ML-Proj-Dataset\\Weather"


excel_files = [f for f in os.listdir(folder_path) if f.endswith('.xlsx')]


excel_files.sort(key=lambda x: int(os.path.splitext(x)[0]))


for file in excel_files:
    
    file_number = int(os.path.splitext(file)[0])
    
    
    year = 2023 if file_number < 7 else 2024
    
    # Process the file
    full_path = os.path.join(folder_path, file)
    process_excel_file(full_path, year)
    print(f"Processed file {file} with year {year}")

print("All files have been processed!")
```

```python
import pandas as pd

def merge_and_show_matches(df_flights, df_weather):
    """
    Merges flight and weather DataFrames and shows matching row numbers
    """
    
    df_flights['Date'] = pd.to_datetime(df_flights['Date'])
    df_weather['TIME'] = pd.to_datetime(df_weather.iloc[:, 0])
    
    
    weather_dates = dict(enumerate(df_weather['TIME']))
    
    
    
    matches = df_flights['Date'].isin(df_weather['TIME'])
    matching_indices = df_flights[matches].index
    
    
    for flight_idx in matching_indices:
        flight_date = df_flights.loc[flight_idx, 'Date']
        weather_idx = df_weather[df_weather['TIME'] == flight_date].index[0]
        print(f"Flight Row {flight_idx + 1} matches Weather Row {weather_idx + 1}")
    
   
    merged_df = pd.merge(
        df_flights,
        df_weather,
        left_on='Date',
        right_on='TIME',
        how='left'
    )
    
    return merged_df

# Usage:
merged_data = merge_and_show_matches(trained_df, weather_df)
```
